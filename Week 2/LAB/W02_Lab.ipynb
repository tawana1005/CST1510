## Topic 1 · Comparisons and Boolean logic

### D1. The six comparison operators

1. Create `a = 10` and `b = 20`.
2. Print the result of each comparison: `a == b`, `a != b`, `a < b`, `a > b`, `a <= b`, `a >= b`

**Expected output:** `False`, `True`, `True`, `False`, `True`, `False` (one per line)

````python
# D1
# compares a and b with each comparison operator where every comparison gives True or False
a = 10      # first number (= asigns value 10 to a)
b = 20      # second number

print(a == b)   # asks if a is equal to b, False
print(a != b)   # asks if a is not equal to b, True
print(a < b)    # asks if a is less than b, True
print(a > b)    # asks if a is greater than b, False
print(a <= b)   # asks if a is less than or equal to b, True
print(a >= b)   # asks if a is greater than or equal to b, False
# == and = are NOT similar
````

### D9. Count from 1 to 10 with `while`

1. Create a variable `count = 1`.
2. Write a `while` loop that runs while `count` is 10 or less.

**Expected output:** the numbers 1 to 10, one per line.

````python
# D9
# counts from 1 to 10 with while
count = 1           # starting value
while count <= 10:  # keep looping while this condition is true
    print(count)   
    count += 1     # add 1 to step up the count
# without the count the loop would not stop
````

### D11. `while True` and `break`: add numbers until the user types `done`

1. Create a variable `total = 0`.
2. Use a `while True:` loop. Inside it, ask: `Enter a number (or done to finish): `
3. If the user types `done`, stop the loop with `break`.
4. Otherwise, convert what they typed with `float()` and add it to `total`.
5. After the loop, print the total.

**Example run:**

```
Enter a number (or done to finish): 5
Enter a number (or done to finish): 10
Enter a number (or done to finish): 2.5
Enter a number (or done to finish): done
Total: 17.5
```

*In a notebook, the input box appears at the top of VS Code. Press Enter after each number.*

````python
# D11
total = 0
while True:
    answer = input("Enter a number (or done to finish):")
    if answer == "done":
        break
    total += float(answer)
# True is always true so the loop contnues until stopped by break
# total += float(answer) keeps adding numbers entered while the loop is running until done 

print("Total =", total)
````

### D12. `continue`: skip one number

Use a `while` loop to print the numbers 1 to 10, but **skip 5** using `continue`.

**Expected output:** `1 2 3 4 6 7 8 9 10` (one per line)

*Hint: add 1 to your counter **before** the `continue`, or the loop gets stuck on 5.*

````python
# D12
count = 0
while count < 10:
    count += 1
    if count == 5:
        continue
    print(count)
# added 1 to the counter before the continue or the loop would be stuck on 5
# continue skips everything below it in that round
# if count += 1 cam after the continue, then when the count reached 5it would jump back to the top without adding 1
````

## Topic 4 · `for` loops and `range()`
*Walkthrough: `04 - for Loops and range`*

### D13. `range(stop)` and `range(start, stop)`

1. Use `for i in range(5):` to print the numbers it gives you.
2. Then write a second loop using `range(start, stop)` that prints 1 to 5.

**Expected output:** `0 1 2 3 4`, then `1 2 3 4 5` (one per line)

*Remember: the `stop` number is never included.*

````python
# D13
# shows how range can work with one number and with two numbers
for i in range(5): # range starts at 0 and stops before 5
    print(i)
for i in range(1,6): # here range starts at 1 and stops before 6 so it prints 1 to 5
    print(i)
````

### D14. `range(start, stop, step)`

1. Print every even number from 2 to 20.
2. Then write a second loop that counts down from 10 to 1.

**Expected output:** `2 4 6 … 20`, then `10 9 8 … 1` (one per line)

*Hint: to count down, use a negative step.*

````python
# D14
for i in range(2,21,2): # starts at 2 and moves by two and stops before 21
    print(i)
for i in range(10,0,-1): # starts at 10 step -1 so it goes from 10 going down and stops before 0
    print(i)
````

### D15. `enumerate()`: the position and the value together

1. Create `code = "A7X"`.
2. Use a `for` loop with `enumerate(code, start=1)` to print each character with its position.

**Expected output:**
```
Character 1: A
Character 2: 7
Character 3: X
```

````python
# D15
"""Walks through each character of code  and shows its position starting at 1"""
code = "A7X"
for i, char in enumerate(code, start=1): # position 1, 2, 3 char: A, 7, X
    print(f"Character {i}: {char}")
````
