### D5. Add a method

1. Add a method `summary(self)` to the class below. It **returns** a string like `R-001: 87.0/100.0`,
   built from `self.label`, `self.value` and `self.limit`.
2. Create a record and print the result of calling `.summary()` on it.

**Expected output:** `R-001: 87.0/100.0`

````python
# D5
class Record:
    """Creates a single record with a label, value and a limit"""
    def __init__(self, label, value, limit):
        """Store the label, value and limit on the objects"""
        self.label = label
        self.value = value
        self.limit = limit

    def summary(self):
        """Return a string like R-001: 87.0/100.0"""
        return f"{self.label}: {self.value}/{self.limit}"   

r1 = Record("R-001", 87.0,100.0)
print(r1.summary()) # the bracket is empty because self is passed automatically so no attribute is needed
````

### D6. `self` is whichever object made the call

1. Copy your class from D5 and add a method `status(self)` that returns `"OVER LIMIT"` (100% or more),
   `"WARNING"` (90% or more) or `"OK"`, worked out from `self.value` and `self.limit`.
2. Create three records and print each one's `.status()`:

| Record | Expected output |
|---|---|
| `Record("R-001", 87.0, 100.0)` | `OK` |
| `Record("R-002", 120.0, 100.0)` | `OVER LIMIT` |
| `Record("R-003", 95.0, 100.0)` | `WARNING` |

````python
# D6
class Record:
    """Creates a single record with a label, value and a limit"""
    def __init__(self, label, value, limit):
        """Store the label, value and limit on the objects"""
        self.label = label
        self.value = value
        self.limit = limit

    def summary(self):
        """Returns a string like R-001: 87.0/100.0"""
        return f"{self.label}: {self.value}/{self.limit}"

    def status(self):
        """Returns OVER LIMIT, WARNING or OK on percent worked out from self.value and self.limit"""
        percent = (self.value / self.limit) * 100   # uses this object's data
        if percent >= 100:
            return "OVER LIMIT"
        elif percent >= 90:
            return "WARNING"
        else:
            return "OK"

r1 = Record("R-001", 87.0, 100.0)
r2 = Record("R-002", 120.0, 100.0)
r3 = Record("R-003", 95.0, 100.0)

print(r1.status())  # OK
print(r2.status())  # OVER LIMIT
print(r3.status())  # WARNING
````

### D7. One method calls another

1. Copy your class from D6 and add a method `percent(self)` that returns `(self.value / self.limit) * 100`.
2. Change `status(self)` so it uses `self.percent()` instead of working out the percentage itself.
3. Create `Record("R-001", 87.0, 100.0)` and print its `.percent()` and `.status()`.

**Expected output:** `87.0`, then `OK`

````python
# D7
class Record:
    """Creates a single record with a label, value and a limit"""
    def __init__(self, label, value, limit):
        """Store the label, value and limit on the objects"""
        self.label = label
        self.value = value
        self.limit = limit

    def summary(self):
        """Returns a string like R-001: 87.0/100.0"""
        return f"{self.label}: {self.value}/{self.limit}"

    def percent(self):
        """Returns percent of value on limit"""
        return (self.value / self.limit) * 100

    def status(self):
        """Returns OVER LIMIT, WARNING or OK on percent worked out from self.value and self.limit"""
        p = self.percent()  # ask this object for its own percent
        if p >= 100:
            return "OVER LIMIT"
        elif p >= 90:
            return "WARNING"
        else:
            return "OK"

    def is_over(self):
        return self.status() == "OVER LIMIT"

r1 = Record("R-001", 87.0, 100.0)
print(r1.percent()) # 87.0
print(r1.status())  # OK
````

### D8. A method can change its own object

1. Define a class `Checker`. In `__init__(self)`, set `self.over_limit_count = 0`.
2. Add a method `check(self, value, limit)` that adds 1 to `self.over_limit_count` when `value` is
   greater than `limit`.
3. Create one `Checker`, call `.check()` three times with `(87, 100)`, `(120, 100)` and `(130, 100)`,
   then print `over_limit_count`.

**Expected output:** `2`

````python
# D8
class Checker:
    """Counts how many values have gone over their limit"""
    def __init__(self):
        """Starts over_limit_count at 0"""
        self.over_limit_count = 0   
    def check(self, value, limit):
        """Adds 1 to the count when value > limit"""
        if value > limit:
            self.over_limit_count += 1

c = Checker()   # calling the class
c.check(87, 100)    # checks if value is over limit, no, so count remains 0
c.check(120,100)    # value is greater than limit so count adds 1 to make it 1
c.check(130, 100)   # value is greater than limit so count adds 1 to make it 2
print(c.over_limit_count)   # 2 values are over limit
````

### D10. A class-level counter

1. Define a new, small class `Counted`. Directly under the `class` line (not inside `__init__`),
   add `count = 0`.
2. In `__init__(self)`, add 1 to `Counted.count`.
3. Create three `Counted` objects, then print `Counted.count`.

**Expected output:** `3`

Add a comment explaining why the count is shared by every object instead of each object having its own.

````python
# D10
class Counted:  # not inside init
    """Counts how many Counted objects have been created"""
    count = 0   # the count is used by all objects
    def __init__(self):
        """Add 1 to the count each time an object is created"""
        Counted.count += 1  # here its not self. but rather Counted.

# Creating 3 counted objects
a = Counted()
b = Counted()
c = Counted()

print(Counted.count)    # 3
# The count belongs to the class and not just one object so every object shares that same count.
# Every time __init__ runs, it adds 1 to the same number
# If each object had its own count (self.count) the every object would start again and 0 so the total will not build up to 3
````

### D11. Load objects from a file

`records_sample.txt` (in this `LAB` folder) has one record per line, like `R-001,87,100`.

1. Write `load_records(path)` that reads the file and returns a list of `Record` objects
   (convert `value` and `limit` with `float()`).
2. Loop over the list and print each record's label and status.

**Expected output:** the same three lines as D9.

````python
# D11
def load_records(path):
    """Reads file containinng label, value, limit and returns a list of Record object"""
    r = []  # starting with an empty list
    with open(path) as f:    
        for line in f:  
            line = line.strip() # remove \n
            if line == "":
                continue
            label, value, limit = line.split(",")   # 3 pieces separated by a ,
            r.append(Record(label, float(value), float(limit)))
    return r    # returns after the loop not inside so i need to do the right indentation

r = load_records("records_sample.txt")

for record in r:
    print(record.label, record.status())
````

### C2. An account that locks

Write a class `Account` with a `username` and a count of failed logins that starts at 0.
It has two methods:

- `fail(self)`: adds 1 to the failed logins
- `locked(self)`: returns `True` if there have been 3 or more failed logins

**Test it:**

| Code | Expected output |
|---|---|
| an account for `"anna"` with `.fail()` called 3 times, then `print(anna.locked())` | `True` |
| an account for `"ben"` with `.fail()` called once, then `print(ben.locked())` | `False` |

````python
# C2
class Account:
    """A user account that locks after three failed logins"""
    def __init__(self, username):
        """Storing the username and starting the failed login count at 0"""
        self.username = username
        self.failed_logins = 0  # start the count at 0

    def fail(self):
        """Function to count failed logins"""
        self.failed_logins += 1

    def locked(self):
        """Retrun True is there have been 3 or more failed logins"""
        if self.failed_logins >= 3:
            return True
        else:
            return False

anna = Account("anna")
anna.fail()
anna.fail()
anna.fail()
print(anna.locked())    # 3 failures, so True

ben = Account("ben")
ben.fail()
print(ben.locked())     # 1 failure, so False
````
