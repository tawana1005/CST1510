## Topic 1 · Case and whitespace
*Walkthrough: `01 - Case and Whitespace`*

### D1. Case changes comparisons

1. Create `a = "SRV-01"` and `b = "srv-01"`.
2. Print `a == b`.
3. Print the same comparison again, with `.lower()` on both sides.

**Expected output:** `False`, then `True`

````python
# D1
# python sees capital and small letters different so for ecample A and a would not be similar
a = "SRV-01"    # capital letters
b = "srv-01"    # small letters
print(a == b)   # different because one is captal letters and one is small so False
print(a.lower() == b.lower())   # .lower makes both to become "srv-01" so the two become equal hence true
````

### D2. `.upper()`, `.lower()`, `.title()`

1. Create `name = "record check console"` and print it with `.title()`, then with `.upper()`.
2. Create `original = "SRV-01"`. Print `original.lower()`, then print `original` again.
3. Add a comment explaining why `original` did not change.

**Expected output:**
```
Record Check Console
RECORD CHECK CONSOLE
srv-01
SRV-01
```

````python
# D2
"""String methods return a new string but the original remains as it is"""
name = "record check console"
print(name.title()) # first letter of each word in capitals
print(name.upper()) # all letters in capital

original = "SRV-01"
print(original.lower()) # displays a new string in lowercase
print(original)     # returns the original unchanged
# original did not change because the string is immutable
# .lower() builds a new string and returns it 
# to keep the new lowercase version, i have to store it as original = original.lower()
````

### D3. `.strip()`

1. Create `raw = "  srv-01  "`.
2. Print it inside square brackets so you can see the spaces: `print(f"[{raw}]")`
3. Print it again the same way, with `.strip()`.

**Expected output:** `[  srv-01  ]`, then `[srv-01]`

````python
# D3
raw = " srv-01 "    # one space before and one space after
print(f"[{raw}]")   # without strip(), the spaces at the ends are printed
print(f"[{raw.strip()}]")  # .strip() returns a new string without the spaces
````

### D4. `.lstrip()`, `.rstrip()` and chaining

1. Create `messy = "***srv-01***"` and print `messy.strip("*")`, `messy.lstrip("*")` and `messy.rstrip("*")`.
2. Create `field = "  SRV-01  "` and print `field.strip().lower()` (two methods chained together).

**Expected output:** `srv-01`, `srv-01***`, `***srv-01`, then `srv-01`

````python
# D4
messy = "***srv-01***"
print(messy.strip("*")) # removes * on both ends
print(messy.lstrip("*"))    # removes * on the left side
print(messy.rstrip("*"))    # removes * on the right side

field = " SRV-01 "
print(field.strip().lower())    # two methods chained together to execute one task
````

## Topic 4 · Parsing real text
*Walkthrough: `04 - Parsing Real Text`*

### D14. Parse one messy line, step by step

Start with `raw_line = "  Srv-01 , used=87 , total=100  "`.

1. `.strip()` the whole line.
2. Split off the hostname with `.split(",", 1)` into `hostname_part` and `rest`.
3. Clean the hostname with `.strip().lower()`.
4. Split `rest` on `","`. For each piece, split on `"="`, strip the key and the value, and convert the
   value with `float()`.
5. Put everything into one dict and print it.

**Expected output:** `{'hostname': 'srv-01', 'used': 87.0, 'total': 100.0}`

````python
# D14
raw_line = " Srv-01 , used=87 , total=100 "

line = raw_line.strip() # strip the whole line
hostname_part, rest = line.split(",", 1)    # split off the hostname
hostname = hostname_part.strip().lower() # cleaning the hostname

record = {"hostname": hostname} # dict is starting with hostname
for piece in rest.split(","):   # one key, value pair at a time
    key, value = piece.split("=")   # cut at "="
    record[key.strip()] = float(value.strip())

print(record)   # printing the dictionary
````

### D15. Turn it into a function

1. Put your steps from D14 into a function `parse_line(raw_line)` that **returns** the dict.
   Give it a one-line docstring.
2. Print `parse_line("SRV-02,used=120,total=100")`.

**Expected output:** `{'hostname': 'srv-02', 'used': 120.0, 'total': 100.0}`

````python
# D15
def parse_line(raw_line):
    """Parse a log line into a dictionary"""
    line = raw_line.strip()
    hostname_part, rest = line.split(",", 1)    # cut at the first comma
    record = {"hostname": hostname_part.strip().lower()}

    for piece in rest.split(","):   # one key, value pair at a time
        key, value = piece.split("=")   # cut at "="
        record[key.strip()] = float(value.strip())

    return(record)  # hand back the dictionary

print(parse_line("SRV-02,used=120,total=100"))  # calling it with the new line and printing the result
````

## Challenge · for a first

### C1. Parse a whole file

Read `messy_lines.txt` (in this `LAB` folder) line by line. Parse each line with `parse_line` from D15,
and print the hostname with `OVER LIMIT` (100% or more), `WARNING` (90% or more) or `OK`.
Do not write the cleaning logic a second time.

**Expected output:**
```
srv-01 OK
srv-02 OVER LIMIT
srv-03 OK
srv-04 WARNING
```

````python
# C1
with open("messy_lines.txt", "r") as f: # open the txt file in read mode
    for line in f:  # read and process the file one line at a time
        record = parse_line(line)   # parse the current line using the parse function from D15

        # extracting individual values from the returned dictionary
        hostname = record["hostname"]
        used = record["used"]
        total = record["total"]

        percent = (used / total) * 100  # calculating the percent

        if percent >= 100:
            status = "OVER LIMIT"
        elif percent >=90:
            status = "WARNING"
        else:
            status = "OK"
        # displaying the hostname and its storage status
        print(hostname, status)
````

### C2. When parsing breaks

Loop over the three lines below. Use `try` / `except ValueError` so that lines that do not parse print
`Could not parse:` and the line, instead of crashing.

```python
lines = ["srv-07,used=50,total=100",
         "srv-08,used=unknown,total=100",
         "srv-09,note=a=b,used=1,total=2"]
```

**Expected output:**
```
{'hostname': 'srv-07', 'used': 50.0, 'total': 100.0}
Could not parse: srv-08,used=unknown,total=100
Could not parse: srv-09,note=a=b,used=1,total=2
```

````python
# C2
lines = ["srv-07,used=50,total=100",
         "srv-08,used=unknown,total=100",
         "srv-09,note=a=b,used=1,total=2"]

for line in lines:
    try:    # attempt to parse line using the D15 function
        record = parse_line(line)
        print(record)   # display the successfully parsed dictionary

    except ValueError:  # handle invalid numerical values without crashing
        print("Could not parse:", line) # phrase and the line that could not parse
````
